```python
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', None)
from pathlib import Path
import pyarrow as pa
import pyarrow.parquet as pq
import re
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
import os, tempfile
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import NearestNeighbors
from STICC_solver import STICC
%matplotlib inline
import matplotlib.pyplot as plt
import glob
```

```python


folder_path = Path("six_fires_parquet/")
parquet_files=[]
for file_path in folder_path.iterdir():
  if file_path.is_file():
    parquet_files.append(file_path.name)
```

```python
features = [
    "total_precipitation",
    "wind_speed",
    "wind_direction",
    "energy_release_component",
    "ndvi",
    "forecast_total_precipitation"
]
def getFireDf(datestr):
    collections, fire_masks = [], []
    fire_id=datestr
    df = pd.read_parquet(f"six_fires_parquet/{fire_id}.parquet")
    dates = sorted(df["date"].unique())
    df = df[df["date"].isin(dates)].copy()
    df["burning"] = df["active_fire"].notna().astype(float)
    df = df.sort_values(["x", "y", "date"])
    return df, dates
            
def plot_fire(df, dates):
    fire_df = df.copy()
    xmin, xmax = fire_df["x"].min(), fire_df["x"].max()
    ymin, ymax = fire_df["y"].min(), fire_df["y"].max()
    
    ncols = 4
    nrows = int(np.ceil(len(dates) / ncols))
    
    fig, axes = plt.subplots(
        nrows,
        ncols,
        figsize=(4*ncols, 4*nrows),
        sharex=True,
        sharey=True,
        squeeze=False
    )
    
    for ax, date in zip(axes.flat, dates):
    
        day = fire_df[fire_df["date"] == date]
    
        ax.scatter(
            day["x"],
            day["y"],
            c=day["active_fire"],
            cmap="bwr",
            vmin=0,
            vmax=1,
            marker="s",
            s=10
        )
    
        ax.set_xlim(xmin, xmax)
        ax.set_ylim(ymin, ymax)
        ax.set_aspect("equal", adjustable="box")
    
        ax.set_title(str(date))
    
    for ax in axes.flat[len(dates):]:
        ax.axis("off")
    
    return fig
    


def sw_dbscan(db, features, eps=.15, min_pts=4, d=1):
    db = db.copy().reset_index(drop=True)
    X = MinMaxScaler().fit_transform(db[features])
    labels = np.full(len(db), -1)
    visited = np.zeros(len(db), bool)
    cid = 0

    def neighbors(i):
        r, c = db.loc[i, ["row", "col"]]
        idx = np.where(
            (abs(db["row"].values - r) <= d) &
            (abs(db["col"].values - c) <= d)
        )[0]
        return idx[np.linalg.norm(X[idx] - X[i], axis=1) <= eps]

    for i in range(len(db)):
        if visited[i]: continue
        visited[i] = True
        N = neighbors(i)
        if len(N) < min_pts: continue

        labels[i] = cid
        Q = list(N)

        while Q:
            j = Q.pop()
            if not visited[j]:
                visited[j] = True
                Nj = neighbors(j)
                if len(Nj) >= min_pts: Q.extend(Nj)
            if labels[j] == -1: labels[j] = cid

        cid += 1

    db["cluster_id"] = labels
    return db   



def sticc_cluster(db, features, k=5, R=3, beta=3, lam=.01, max_iters=20):
    db = db.dropna(subset=features).copy().reset_index(drop=True)

    if len(db) < max(k, R + 1):
        db["cluster_id"] = -1
        return db

    X = StandardScaler().fit_transform(db[features])

    # R nearest spatial neighbors, excluding the cell itself
    R_eff = min(R, len(db) - 1)

    nn = NearestNeighbors(n_neighbors=R_eff)
    nn.fit(db[["row", "col"]])
    
    neighbors = nn.kneighbors(return_distance=False)

    data = np.hstack([
        np.arange(len(db))[:, None],
        X,
        neighbors
    ])

    with tempfile.NamedTemporaryFile(suffix=".txt", delete=False) as f:
        fname = f.name

    np.savetxt(fname, data, delimiter=",")

    os.makedirs("sticc_output", exist_ok=True)

    model = STICC(
        spatial_radius=R,
        number_of_clusters=k,
        lambda_parameter=lam,
        beta=beta,
        maxIters=max_iters,
        threshold=2e-5,
        write_out_file=False,
        prefix_string="sticc_output/",
        num_proc=1,
        biased=True,
        attr_idx_start=1,
        attr_idx_end=len(features),
        spatial_idx_start=len(features)+1,
        spatial_idx_end=len(features)+R
    )

    labels, mrfs = model.fit(input_file=fname)
    os.remove(fname)

    db["cluster_id"] = np.asarray(labels).astype(int)
    return db
```

```python
df=getFireDf("2018_fire_21615469")[0]
one_day=df[df['date']=='2018-03-11']
one_day
burning = one_day[one_day["burning"] == 1.0].copy()

clustered = sw_dbscan(
    burning,
    features=features,
    eps=0.2,
    min_pts=4,
    d=1
)
print(clustered["cluster_id"].value_counts().sort_index())
print("Clusters:", clustered.loc[clustered.cluster_id != -1, "cluster_id"].nunique())
plt.figure(figsize=(8, 8))

sc = plt.scatter(
    clustered["x"],
    clustered["y"],
    c=clustered["cluster_id"],
    cmap="tab20",
    marker="s",
    s=20
)

plt.colorbar(sc, label="Cluster ID")
plt.gca().set_aspect("equal")
plt.title(str(clustered["date"].iloc[0]))
plt.show()
```

```python
xmin, xmax = df["x"].min(), df["x"].max()
ymin, ymax = df["y"].min(), df["y"].max()
K = 3
R = 3
dates = sorted(df["date"].unique())
valid_dates = []

for date in dates:
    day = df[df["date"] == date]
    burning = day[day["burning"] == 1]

    if len(burning) >= max(K, R + 1):
        valid_dates.append(date)


ncols = 2
nrows = int(np.ceil(len(valid_dates) / ncols))

fig, axes = plt.subplots(
    nrows, ncols,
    figsize=(8*ncols, 8*nrows),
    sharex=True,
    sharey=True,
    squeeze=False
)

for ax, date in zip(axes.flat, valid_dates):

    day = df[df["date"] == date].copy()
    burning = day[day["burning"] == 1].copy()

    if len(burning) < max(K, R + 1):
        ax.axis("off")
        continue

    clustered = sticc_cluster(
        burning,
        features=features,
        k=K,
        R=R,
        beta=3,
        lam=.01,
        # verbose=False
    )

    ax.scatter(
        clustered["x"],
        clustered["y"],
        c=clustered["cluster_id"],
        cmap="tab20",
        marker="s",
        s=15,
        vmin=0,
        vmax=K-1
    )

    ax.set_xlim(xmin, xmax)
    ax.set_ylim(ymin, ymax)
    ax.set_aspect("equal")
    ax.set_title(str(date))

for ax in axes.flat[len(valid_dates):]:
    ax.axis("off")

plt.tight_layout()
display(fig)
```

```python
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score

def cluster_metrics(X, labels):
    if len(set(labels)) < 2:
        return None

    return {
        "silhouette": silhouette_score(X, labels),
        "calinski_harabasz": calinski_harabasz_score(X, labels),
        "davies_bouldin": davies_bouldin_score(X, labels)
    }
```

```python
results = []

for file in glob.glob("six_fires_parquet/*.parquet"):
    fire_id = os.path.basename(file).replace(".parquet", "")
    df, dates = getFireDf(fire_id)
    for date in dates:
        day = df[df["date"] == date]
        burning = day[day["burning"] == 1].copy()

        if len(burning) < 2:
            continue

        clustered = sw_dbscan(
            burning,
            features=features,
            eps=0.2,
            min_pts=4,
            d=1
        )

        valid = clustered[clustered["cluster_id"] != -1].copy()

        if valid["cluster_id"].nunique() < 2:
            continue

        X = MinMaxScaler().fit_transform(valid[features])
        labels = valid["cluster_id"].to_numpy()

        m = cluster_metrics(X, labels)

        results.append({
            "fire_id": fire_id,
            "date": date,
            **m
        })

sw_metrics = pd.DataFrame(results)
```

```python
results = []

for file in glob.glob("six_fires_parquet/*.parquet"):
    
    fire_id = os.path.basename(file).replace(".parquet", "")
    print(fire_id)
    df, dates = getFireDf(fire_id)

    for date in dates:

        day = df[df["date"] == date]
        burning = day[day["burning"] == 1].copy()

        if len(burning) < max(K, R + 1):
            continue

        clustered = sticc_cluster(
            burning,
            features=features,
            k=K,
            R=R,
            beta=3,
            lam=.01
        )

        if clustered["cluster_id"].nunique() < 2:
            continue

        X = StandardScaler().fit_transform(clustered[features])
        labels = clustered["cluster_id"].to_numpy()

        m = cluster_metrics(X, labels)

        results.append({
            "fire_id": fire_id,
            "date": date,
            **m
        })

sticc_metrics = pd.DataFrame(results)
```

```python
sticc_metrics
```

```python
comparison = sw_metrics.merge(
    sticc_metrics,
    on=["fire_id", "date"],
    suffixes=("_sw", "_sticc")
)
comparison
```

```python
comparison["silhouette_diff"] = (
    comparison["silhouette_sticc"] - comparison["silhouette_sw"]
)

comparison["ch_diff"] = (
    comparison["calinski_harabasz_sticc"]
    - comparison["calinski_harabasz_sw"]
)

comparison["db_diff"] = (
    comparison["davies_bouldin_sw"]
    - comparison["davies_bouldin_sticc"]
)
```
